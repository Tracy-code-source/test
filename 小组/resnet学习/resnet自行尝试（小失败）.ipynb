```python
import torch
from torch import nn
import torchvision
from torchvision import models
from torchvision import datasets, transforms
```

```python
# 训练
model = models.resnet18(weights = None)
num_features = model.fc.in_features
model.fc = nn.Linear(num_features, 10) # 十分类问题
```

```python
model
```

```python
from torchsummary import summary
```

```python

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# 将模型转移到GPU上
model.to(device)
```

```python
#summary(model, input_size = (3, 224, 224), device = 'cuda')
```

```python
# 批量读取
batch_size = 64
```

```python
device
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),         #将numpy类型转换为Tensor张量
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))    #归一化预处理，加速模型训练
])
```

```python
# 下载训练集及测试集
trainset = datasets.CIFAR10(root = './CIFAR10', train = True, download = True, transform = transform)
testset = datasets.CIFAR10(root = './CIFAR10', train = False, download = True, transform = transform)
```

```python
train_dataloader = torch.utils.data.DataLoader(dataset = trainset, shuffle = True, batch_size = batch_size)
test_dataloader = torch.utils.data.DataLoader(dataset = testset, shuffle = True, batch_size = batch_size)
```

```python
images, labels = next(iter(train_dataloader))
```

```python
images.shape
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr = 0.001, momentum = 0.9)
total_batch = len(train_dataloader)
```

```python
for i in range(20):
    for j, (images, labels) in enumerate(train_dataloader):
        images = images.to(device)
        labels = labels.to(device)
        outputs = model(images)
        loss = criterion(outputs, labels)
        # 标准化处理
        n_corrects = (outputs.argmax(axis = 1) == labels).sum().item()  # 找到预测类型并和标签比较
        acc = n_corrects / labels.size(0)
        # backward
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        if (j + 1) % 100 == 0:
            print("Loss = {:.3f}, Accuracy : {:.3f}".format(loss.item(), acc))
        
```

```python
# 模型评估
total = 0
correct = 0
for images, labels in  test_dataloader:
    images = images.to(device)
    labels = labels.to(device)
    outputs = model(images)
    predicted = torch.argmax(outputs, dim = 1)
    total += images.size(0)
    correct += (predicted == labels).sum().item()
print("Accuracy : {:.3f}".format(correct / total))
```
